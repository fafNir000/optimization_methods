```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
```

```python
def func(x):
  return (1/2)*(x**4)+2*x*x-8*x+6
```

```python
a = 0
b = 2
eps = 0.5
n = 8
x = np.array([])
fx = np.array([])
for i in range(int(n)+1):
  x = np.append(x, a+i*((b-a)/n))
  fx = np.append(fx, func(a+i*((b-a)/n)))
```

```python
rx = np.linspace(a, b, num=100)
frx = func(rx)

res = optimize.minimize(func, x0=0.)
mini = res.fun
xmin = res.x[0]
```

```python
plt.plot(rx, frx)
plt.scatter(x, fx, color='red', s=25)
plt.scatter(x[np.argmin(fx)], fx.min(),color='green', s=25)
plt.scatter(xmin, mini,color='violet', s=25)
plt.grid(True)
plt.show()
```

```python
print(f"function real minimum :: ({xmin}, {mini})")
print(f"function minimum :: ({x[np.argmin(fx)]}, {fx.min()})")
print(f"X difference :: {abs(x[np.argmin(fx)]-xmin)} < {eps}")
```
