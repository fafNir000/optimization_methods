```python
import numpy as np
```

```python
weights = np.array([15,40,45])
prob = np.array([0.45,0.45,0.1])
```

```python
s=0
n = 10
for i in range(n):
  rec = np.random.choice(weights,p=prob)
  inp = input(f"Do you want to take {rec}? (y/n) :: ")
  if inp == "y":
    s=rec
    break
print(s)
```

```python
v = np.array(weights,dtype=float).reshape(1, -1)
expectation = np.array([0],dtype=float)
for i in range(n-1):
  exp = v[0] @ prob
  v = np.insert(v, 0, [max(w,exp) for w in weights], axis=0)
  expectation = np.insert(expectation, 0, exp)
```

```python
print("----------------------------------")
for i in range(len(v)):
  print(f"Room {i+1} :: Expectation {expectation[i]}")
  print(f"V{i+1} :: {v[i]}")
```

```python
s=0
n = 10
for i in range(n):
  rec = np.random.choice(weights,p=prob)
  print(f"Room {i+1}")
  print(f"Do you want to take {rec}?")
  if rec >= expectation[i]:
    s=rec
    print("Answer :: Yes")
    break
  else:
    print("Answer :: No")
print(s)
```
