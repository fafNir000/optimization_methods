```python
import numpy as np
```

```python
def func(x):
  return 0.5*(x[0]**2+2*x[1]**2+3*x[2]**2)-(x[0]+2*x[1]+3*x[2])
def defuncx1(x):
  return x[0]-1
def defuncx2(x):
  return 2*(x[1]-1)
def defuncx3(x):
  return 3*(x[2]-1)
```

```python
x0 = np.array([0,0,0])
eps = 0.2
x = x0.copy()
lr = 0.1
iter = 10
for k in range(iter):
  grad = np.array([defuncx1(x), defuncx2(x), defuncx3(x)])
  lr = ((grad[0])**2+(grad[1])**2+(grad[2])**2)/((grad[0])**2+2*(grad[1])**2+3*(grad[2])**2)
  vecsize = np.linalg.norm(grad)
  xstr = [f"{i:.4f}" for i in x]
  print(f"Iteration :: {k}, Point :: {xstr}, Learning rate :: {lr}, Gradient norm :: {vecsize}")
  if vecsize < eps:
    print(f"Function minimum :: {func(x)}")
    break
  x = x - lr*grad
```
