```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
```

```python
def func(x):
  return (1/2)*(x**4)+2*x*x-8*x+6
```

```python
a = 0
b = 2
eps = 0.5
x1 = a + (3-np.sqrt(5))*(b-a)/2
x2 = a + (np.sqrt(5)-1)*(b-a)/2
avgx = (a+b)/2
f1 = func(x1)
f2 = func(x2)
f = func(avgx)
arrxavg = np.array([avgx])
arrf = np.array([f])
while abs(a-b)>=eps:
  if (f1<=f2):
    b = x2
    x2 = x1
    x1 = a + b - x1
  else:
    a = x1
    x1 = x2
    x2 = a + b - x2
  avgx = (a+b)/2
  f1 = func(x1)
  f2 = func(x2)
  f = func(avgx)
  arrxavg = np.append(arrxavg,avgx)
  arrf = np.append(arrf,f)
```

```python
a = 0
b = 2
rx = np.linspace(a, b, num=100)
frx = func(rx)
res = optimize.minimize(func, x0=0.)
mini = res.fun
xmin = res.x[0]
```

```python
plt.plot(rx, frx)
plt.scatter(arrxavg, arrf, color='black', s=25)
plt.scatter(xmin, mini,color='violet', s=25)
plt.grid(True)
plt.show()
```

```python
print(f"function real minimum :: ({xmin}, {mini})")
print(f"function minimum :: ({arrxavg[np.argmin(arrf)]}, {arrf.min()})")
print(f"X difference :: {abs(arrxavg[np.argmin(arrf)]-xmin)} < {eps}")
```
