```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
```

```python
def func(x):
  return (1/2)*(x**4)+2*x*x-8*x+6
```

```python
a = 0
b = 2
eps = 0.5
delt = 0.125
arrxavg = np.array([(a+b)/2])
arrfxavg = np.array([func((a+b)/2)])

while abs(a-b)>eps:
  x1 = (a+b-delt)/2
  x2 = (a+b+delt)/2
  xavg = (a+b)/2
  f1 = func(x1)
  f2 = func(x2)
  favg = func(xavg)
  arrxavg = np.append(arrxavg, xavg)
  arrfxavg = np.append(arrfxavg,favg)
  if f1<=f2:
    b = x2
  else:
    a = x1
```

```python
a = 0
b = 2
rx = np.linspace(a, b, num=100)
frx = func(rx)
res = optimize.minimize(func, x0=0.)
mini = res.fun
xmin = res.x[0]
```

```python
plt.plot(rx, frx)
plt.scatter(arrxavg, arrfxavg, color='black', s=25)
plt.scatter(xmin, mini,color='violet', s=25)
plt.grid(True)
plt.show()
```

```python
print(f"function real minimum :: ({xmin}, {mini})")
print(f"function minimum :: ({arrxavg[np.argmin(arrfxavg)]}, {arrfxavg.min()})")
print(f"X difference :: {abs(arrxavg[np.argmin(arrfxavg)]-xmin)} < {eps}")
```
